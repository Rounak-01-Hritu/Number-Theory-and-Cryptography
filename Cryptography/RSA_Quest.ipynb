```python
import sys
import math

sys.setrecursionlimit(10**7)

def ConvertToInt(message_str):
    result = 0
    for character in message_str:
        result = result * 256 + ord(character)
    return result

def ConvertToStr(number):
    if number == 0:
        return ""
    result = []
    while number > 0:
        result.append(chr(number % 256))
        number //= 256
    return "".join(reversed(result))

def PowMod(a, n, modulo):
    # Fast modular exponentiation: O(log n)
    result = 1
    a %= modulo
    while n > 0:
        if n % 2 == 1:
            result = (result * a) % modulo
        a = (a * a) % modulo
        n //= 2
    return result

def ExtendedEuclid(a, b):
    # Returns (x, y) such that a*x + b*y = gcd(a, b)
    if b == 0:
        return (1, 0)
    x1, y1 = ExtendedEuclid(b, a % b)
    return (y1, x1 - (a // b) * y1)

def InvertModulo(a, n):
    x, y = ExtendedEuclid(a, n)
    if (a * x + n * y) != 1:
        raise ValueError("The modular inverse does not exist.")
    return x % n

def Decrypt(ciphertext, p, q, exponent):
    modulo = p * q
    phi = (p - 1) * (q - 1)
    private_exponent = InvertModulo(exponent, phi)
    message_as_int = PowMod(ciphertext, private_exponent, modulo)
    return ConvertToStr(message_as_int)

p = 779849711281
q = 748173698927
e = 1018651
ciphertext = 148784435264686331994392
decrypt_first_puzzle = Decrypt(ciphertext, p, q, e)
print("First puzzle:", decrypt_first_puzzle)
```

```python
def ChineseRemainderTheorem(n1, r1, n2, r2):
    # Solve x = r1 (mod n1), x = r2 (mod n2), assuming coprime moduli.
    inverse = InvertModulo(n1, n2)
    multiplier = ((r2 - r1) * inverse) % n2
    return (r1 + n1 * multiplier) % (n1 * n2)

def DecipherHastad(first_ciphertext, first_modulo,
                   second_ciphertext, second_modulo):
    # This two-recipient version uses exponent e=2.
    combined = ChineseRemainderTheorem(
        first_modulo, first_ciphertext,
        second_modulo, second_ciphertext
    )
    message_as_int = math.isqrt(combined)
    if message_as_int * message_as_int != combined:
        raise ValueError("CRT result is not a perfect square; check the supplied values.")
    return ConvertToStr(message_as_int)

# Fill these four values from the second secret document.
first_ciphertext = 1695925818934701958536732606471007147871697773728881305981426383004125888187531096540654444788889942827682032651915812560
first_modulo = 2019703097262944143745237345710626601616589344196612447233041278945420303895066774752152001687355425741826173570952817341
second_ciphertext = 438180481071044087096048173519586369091293503891810459558098521655659500752405553498472135529822589328291372879622802017
second_modulo = 968914506353797853274340817936111164236562754898105214422095149641604042531040243209810736059746240506447210425767208577

values = (first_ciphertext, first_modulo, second_ciphertext, second_modulo)
if None not in values:
    decrypt_seventh_puzzle = DecipherHastad(*values)
    print("Seventh puzzle:", decrypt_seventh_puzzle)
else:
    print("Enter both ciphertexts and moduli from the second secret document.")
```
